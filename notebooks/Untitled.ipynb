```python
model_df = df[
    [
        "Date",
        "Time",
        "Day",
        "Line",
        "Station",
        "Code",
        "Bound",
        "Min Delay",
    ]
].copy()
```

```python
model_df["Date"] = pd.to_datetime(model_df["Date"])
model_df["month"] = model_df["Date"].dt.month
```

```python
model_df["hour"] = model_df["Time"].str.split(":").str[0].astype(int)
```

```python
model_df["Line"] = model_df["Line"].fillna("UNKNOWN")
model_df["Bound"] = model_df["Bound"].fillna("UNKNOWN")
```

```python
model_df = model_df.drop(columns=["Date", "Time"])
```

```python
X = model_df.drop(columns=["Min Delay"])
y = model_df["Min Delay"]
```

```python
timestamp = pd.to_datetime(
    df["Date"].str[:10] + " " + df["Time"]
)

sorted_idx = timestamp.sort_values().index
```

```python
n = len(sorted_idx)

train_end = int(0.70 * n)
val_end = int(0.85 * n)

train_idx = sorted_idx[:train_end]
val_idx = sorted_idx[train_end:val_end]
test_idx = sorted_idx[val_end:]
```

```python
X_train = X.loc[train_idx]
y_train = y.loc[train_idx]

X_val = X.loc[val_idx]
y_val = y.loc[val_idx]

X_test = X.loc[test_idx]
y_test = y.loc[test_idx]
```

```python
print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)
```

```python
from sklearn.dummy import DummyRegressor
from sklearn.metrics import mean_absolute_error

baseline = DummyRegressor(strategy="median")

baseline.fit(X_train, y_train)

val_predictions = baseline.predict(X_val)

baseline_mae = mean_absolute_error(y_val, val_predictions)

print("Baseline MAE:", baseline_mae)
```

```python
categorical_features = [
    "Day",
    "Line",
    "Station",
    "Code",
    "Bound",
]

numeric_features = [
    "month",
    "hour",
]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features,
        ),
        (
            "numeric",
            "passthrough",
            numeric_features,
        ),
    ]
)
```

```python
from sklearn.linear_model import Ridge
from sklearn.pipeline import Pipeline

model = Pipeline(
    steps=[
        ("preprocessing", preprocessor),
        ("regressor", Ridge(alpha=1.0)),
    ]
)
```

```python
model.fit(X_train, y_train)
```

```python
val_predictions = model.predict(X_val)

ridge_mae = mean_absolute_error(y_val, val_predictions)

print("Ridge MAE:", ridge_mae)
```

```python
from xgboost import XGBRegressor

xgb_model = Pipeline(
    steps=[
        ("preprocessing", preprocessor),
        (
            "regressor",
            XGBRegressor(
                n_estimators=400,
                max_depth=6,
                learning_rate=0.05,
                subsample=0.8,
                colsample_bytree=0.8,
                random_state=42,
                n_jobs=-1,
            ),
        ),
    ]
)
```

```python
xgb_model.fit(X_train, y_train)
```

```python
xgb_predictions = xgb_model.predict(X_val)

xgb_mae = mean_absolute_error(y_val, xgb_predictions)

print("Baseline MAE:", baseline_mae)
print("Ridge MAE:", ridge_mae)
print("XGBoost MAE:", xgb_mae)
```

```python
test_predictions = xgb_model.predict(X_test)

test_mae = mean_absolute_error(y_test, test_predictions)

print("Validation MAE:", xgb_mae)
print("Test MAE:", test_mae)
```
